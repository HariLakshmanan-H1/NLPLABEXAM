IMPORTS

```python
from gensim.models import Word2Vec
import numpy as np
from nltk import word_tokenize,sent_tokenize
import pandas as pd
from nltk.util import bigrams
```

LOADING DATA

```python
with open("sherlockholmes.txt", "r", encoding="utf-8") as f:
    text = f.read()

sentences = sent_tokenize(text)

df = pd.DataFrame({"Review": sentences})
```

TOKENIZATION

```python
tokens = [
    [
        w
        for w in word_tokenize(sentence.lower())
        if w.isalpha()
    ]
    for sentence in df['Review']
]
```

MAKING BIGRAMS

```python
bigram_tokens = []

for sentence in tokens:
    sentence_bigrams = [
        "_".join(bg)
        for bg in bigrams(sentence)
    ]
    bigram_tokens.append(sentence_bigrams + ["<UNK>"])
```

MODEL

```python
cbow = Word2Vec(bigram_tokens,window=5,vector_size=100,min_count=1,sg=0)
skipgram = Word2Vec(bigram_tokens,window=5,vector_size=100,min_count=1,sg=1)
```

WORD PAIR

```python
word_pairs = [("sherlock_holmes","sherlock_david",)]
```

COSINE SIMILARITY

```python
def cosine_similarity(w1,w2,model):

    if w1 not in model.wv:
        w1 = "<UNK>"
    
    if w2 not in model.wv:
        w2 = "<UNK>"

    v1 = model.wv[w1] 
    v2 = model.wv[w2] 

    return np.dot(v1,v2) / (np.linalg.norm(v1)*np.linalg.norm(v2))

    
```

COMPUTING COSINE SIMILARITY

```python
for w1,w2 in word_pairs:
    sim1 = cosine_similarity(w1,w2,cbow)
    sim2 = cosine_similarity(w1,w2,skipgram)

    print(f"CBOW similarity between '{w1}' and '{w2}': {sim1:.4f}")
    print(f"Skip-gram similarity between '{w1}' and '{w2}': {sim2:.4f}")
```
