```python
import nltk
from nltk.parse import InsideChartParser,ViterbiParser
from nltk import PCFG
```

```python
pcfg_grammar = PCFG.fromstring(""" 
S -> NP VP [1.0]
VP -> V NP [0.7] | VP PP [0.3]
PP -> P NP [1.0]
NP -> NP PP [0.4] | 'astronomers' [0.1] | 'stars' [0.18] | 'ears' [0.18] | 'saw' [0.04] | 'telescopes' [0.1]
V -> 'saw' [1.0]
P -> 'with' [1.0]
""")
```

```python
sentence_tokens = ['astronomers', 'saw', 'stars', 'with', 'ears']
```

```python
inside_chart_parser = InsideChartParser(pcfg_grammar)
parse_trees = inside_chart_parser.parse(sentence_tokens)
```

```python
total_prob = 0
```

```python
for idx,tree in enumerate(parse_trees,start=1):
    print(f"Parse Tree #{idx}")
    tree.pretty_print()
    print("Probability:", tree.prob())
    total_prob += tree.prob()
```

```python
print(f"Total Probability of all parse trees: {total_prob}")
```

```python
viterbi_parser = ViterbiParser(pcfg_grammar)
```

```python
for best_tree in viterbi_parser.parse(sentence_tokens):
    print("Best Parse Tree:")
    best_tree.pretty_print()
    print("Probability of Best Parse Tree:", best_tree.prob())
```
