IMPORTS

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

LOADING DATA

```python
df = pd.read_csv(
    'moviereview.txt',
    sep='\|\|',
    engine="python",
    names=['review','sentiment'],
    skipinitialspace=True
)
```

LABEL ENCODING

```python
le = LabelEncoder()
df['label'] = le.fit_transform(df['sentiment'])
print("Label Encoding :",dict(
    zip(
        le.classes_,
        le.transform(le.classes_)
    )
))
```

TRAIN TEST SPLIT

```python
X_train,X_test,y_train,y_test = train_test_split(
    df['review'],
    df['label'],
    test_size=0.2,
    random_state=132
)
```

BOW

```python
bow = CountVectorizer()
X_train_bow = bow.fit_transform(X_train)
X_test_bow = bow.transform(X_test)
```

```python
model_bow = MultinomialNB()
model_bow.fit(X_train_bow,y_train)
y_pred_bow = model_bow.predict(X_test_bow)
```

TF-IDF

```python
tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
model_tfidf = MultinomialNB()
model_tfidf.fit(X_train_tfidf,y_train)
y_pred_tfidf = model_tfidf.predict(X_test_tfidf)
```

ACCURACY

```python
print("ACCURACY SCORES")
print(f"BOW - {accuracy_score(y_test,y_pred_bow)}")
print(f"TF-IDF - {accuracy_score(y_test,y_pred_tfidf)}")
```

CLASSIFICATION REPORT

```python
print("CLASSIFICATION REPORT")
print("BOW")
print(classification_report(
    y_test,
    y_pred_bow,
    target_names = le.classes_
))

print("TF-IDF")
print(classification_report(
    y_test,
    y_pred_tfidf,
    target_names = le.classes_
))
```

CONFUSION MATRIX

```python
print("CONFUSION MATRIX")
print("BOW")
print(confusion_matrix(
    y_test,
    y_pred_bow
),end="\n\n")

print("TF-IDF")
print(confusion_matrix(
    y_test,
    y_pred_tfidf
))
```

ACTUAL VS PREDICTED

```python
print("Actual vs Predicted")
for i in range(len(y_test)):
    actual = le.inverse_transform(
                                    [
                                        y_test.iloc[i]
                                    ]
                                )[0]
    bow_pred = le.inverse_transform(
                                    [
                                        y_pred_bow[i]
                                    ]
                                )[0]
    tfidf_pred = le.inverse_transform(
                                    [
                                        y_pred_tfidf[i]
                                    ]
                                )[0]
    
    print(
        f"Actual: {actual:<8} | "
        f"BoW: {bow_pred:<8} | "
        f"TF-IDF: {tfidf_pred:<8}"
    )
```
