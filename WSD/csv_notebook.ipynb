```python
import pandas as pd
from collections import Counter
from math import log2
import numpy as np
```

```python
df = pd.read_csv(r"set2.csv")
df
```

```python
df_music = df[df["Class"]=="Music"]
df_sports = df[df["Class"]=="Sports"]
df_sales = df[df["Class"]=="Sales"]
df_test = df[df["Class"]=="?"]
```

```python
music_count = Counter([
                        w.lower()  
                        for words in df_music["Words"] 
                        for w in words.split(', ') 
                    ])
sports_count = Counter([
                        w.lower()  
                        for words in df_sports["Words"] 
                        for w in words.split(', ') 
                    ])

sales_count = Counter([
                        w.lower()  
                        for words in df_sales["Words"] 
                        for w in words.split(', ') 
                    ])

```

```python
p_music = len(df_music) / len(df)
p_sports = len(df_sports) / len(df)
p_sales = len(df_sales) / len(df)
```

```python
V = len(
    set(music_count) |
    set(sports_count) |
    set(sales_count)
)
```

```python
def classify(words):
    log_music = log2(p_music)
    log_sales = log2(p_sales)
    log_sports = log2(p_sports)

    for w in words.split(", "):
        log_music += log2(
                            (
                                music_count.get(w,0) + 1 
                            ) / (
                                sum(music_count.values()) + V
                            )
                        )
        log_sports += log2(
                            (
                                sports_count.get(w,0) + 1
                            )/(
                                sum(sports_count.values()) + V
                            )
                        )
        log_sales += log2(
                            (
                                sales_count.get(w,0) + 1
                            )/(
                                sum(sales_count.values()) + V
                            )
                        )
        
    scores = {
        "Music" : log_music,
        "Sports" : log_sports,
        "Sales" : log_sales
    }

    return max(scores,key=scores.get)
```

```python
for words in df_test["Words"]:
    print(words,"->",classify(words))
```
