IMPORTS

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
from tensorflow.keras import Sequential
from tensorflow.keras.layers import SimpleRNN,Dense
```

TRAINING SAMPLES

```python
eng = ["hello","bye"]
fra = ["bonjour","salut"]
```

SET OF CHARACTERS

```python
chars = sorted(
    list(
        set(
            " ".join(eng+fra)
        )
    )
)
```

ENCODING AND DECODING

```python
char_to_int = {c:i for i,c in enumerate(chars)}
int_to_char = {i:c for c,i in char_to_int.items()}
```

```python
X = [[char_to_int[c] for c in word] for word in eng]
Y = [[char_to_int[c] for c in word] for word in fra]
```

SETTING VECTOR SIZE

```python
max_len = max(
    max(len(x) for x in X),
    max(len(y) for y in Y)
)
```

PADDING SEQUENCES

```python
X = pad_sequences(
    X,
    maxlen = max_len,
    padding = "post"
)

Y = pad_sequences(
    Y,
    maxlen = max_len,
    padding = "post"
)
```

CREATING ONE SHOT REPRESENTATION

```python
Y = to_categorical(Y,num_classes = len(chars))
```

CREATING MODEL

```python
model = Sequential([
    SimpleRNN(
        32,
        return_sequences = True,
        input_shape = (max_len,1)
        ),
    Dense(
        len(chars),
        activation = 'softmax'
        )
])
```

```python
model.compile(
    optimizer = "adam",
    loss = "categorical_crossentropy"
)
```

TRAINING MODEL

```python
model.fit(
            X,
            Y,
            epochs=300,
            verbose=1
    )
```

TRANSLATION

```python
def translate(word):
    seq = [char_to_int[ch] for ch in word]
    pad = pad_sequences(
                        [seq],
                        maxlen = max_len,
                        padding = "post"    
        )
    out = model.predict(pad)[0]
    return "".join(int_to_char[np.argmax(v)] for v in out)
```

```python
print("hello ->", translate("hello"))
print("bye ->", translate("bye"))
```
