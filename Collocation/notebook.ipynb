IMPORTS

```python
from nltk import word_tokenize,FreqDist,bigrams
from nltk.corpus import stopwords
import math
from math import sqrt
import pandas as pd
```

```python
# text = """
# Rama went to the forest. Sita went with Rama.
# Rama fought the demon. The demon feared Rama.
# """
```

LOADING DATASET

```python
df = pd.read_excel("SASTRA University.xlsx",header=None)
text = "".join(df[0].astype(str))
text
```

STOPWORD REMOVAL AND TOKENIZATION

```python
def preprocess(text):
    stop = set(stopwords.words('english'))
    tokens = word_tokenize(text.lower())
    return [w for w in tokens if w.isalpha() and w not in stop]
```

t - TEST

```python
def t_test(w1,w2,wf,bgf,N):
    O = bgf[(w1,w2)]
    if O == 0:
        return 0
    E = (wf[w1]*wf[w2]) / N

    return (O-E) / sqrt(O)
```

CHI - SQUARE

```python
def chi_square(w1,w2,wf,bgf,N):
    O = bgf[(w1,w2)]
    if O == 0:
        return 0
    E = (wf[w1]*wf[w2]) / N

    return ((O-E)**2)/E
```

INITIALIZATION

```python
words = preprocess(text)
wf = FreqDist(words)
bg = list(bigrams(words))
bgf = FreqDist(bg)
N = len(bg)
```

EVALUATION

```python
def evaluate(test_func,name):
    print(f"{name} scores:")
    for w1,w2 in bgf:
        score = round(test_func(w1,w2,wf,bgf,N),4)
        if test_func == t_test:
            print((w1, w2), ":", score,
              "✓" if score > 2 else "✗")
        else:
            print((w1, w2), ":", score,
              "✓" if score > 3.84 else "✗")
```

```python
evaluate(t_test, "T-Score")
evaluate(chi_square, "Chi-Square")
```
