IMPORTS

```python
from nltk import word_tokenize,FreqDist,bigrams
from nltk.corpus import stopwords
import math
from math import sqrt
import pandas as pd
```

LOADING DATASET

```python
df = pd.read_excel("SASTRA University.xlsx",header=None)
text = "".join(df[0].astype(str))
text
```

STOPWORD REMOVAL AND TOKENIZATION

```python
def preprocess(text):
    stop = set(stopwords.words('english'))
    tokens = word_tokenize(text.lower())
    return [w for w in tokens if w.isalpha() and w not in stop]
```

t - TEST

```python
def t_test(w1,w2,wf,bgf,N):
    # t = () x_bar - mu ) / se
    # se = s^2 / N
    # s^2 = x_bar * (1 - x_bar)

    x_bar = bgf[(w1,w2)] / (N-1)
    mu = (wf[w1]/N) * (wf[w2]/N)

    s2 = x_bar * (1-x_bar)
    se = sqrt(s2/N)

    if se == 0:
        return 0
    
    return (x_bar - mu) / se
```

CHI - SQUARE

```python
def chi_square(w1,w2,wf,bgf,N):
    O11 = bgf[(w1,w2)]
    O12 = wf[w1] - O11
    O21 = wf[w2] - O11
    O22 = N - (O11 + O12 + O21)

    total_O = O11 + O12 + O21 + O22

    E11 = (wf[w1]*wf[w2]) / N
    E12 = (wf[w1]*(N-wf[w2])) / N
    E21 = ((N - wf[w1])*wf[w2]) / N
    E22 = ((N-wf[w1])*(N-wf[w2])) / N

    return (((O11 - E11)**2 / E11) + ((O12 - E12)**2 / E12) + ((O21 - E21)**2 / E21) + ((O22 - E22)**2 / E22))
```

INITIALIZATION

```python
words = preprocess(text)
wf = FreqDist(words)
bg = list(bigrams(words))
bgf = FreqDist(bg)
N = len(bg)
```

EVALUATION

```python
def evaluate(test_func,name):
    print(f"{name} scores:")
    for w1,w2 in bgf:
        score = round(test_func(w1,w2,wf,bgf,N),4)
        if test_func == t_test:
            print((w1, w2), ":", score,
              "✓" if score > 2 else "✗")
        else:
            print((w1, w2), ":", score,
              "✓" if score > 3.84 else "✗")
```

```python
evaluate(t_test, "T-Score")
evaluate(chi_square, "Chi-Square")
```
